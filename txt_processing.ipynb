```python
import re

def find_uncommon_special_characters(file_path):
    """
    读取文件并查找非字母数字、非空白、非常见标点符号的特殊字符及其 Unicode 码点和位置。

    :param file_path: 文本文件的路径
    """
    # 定义要排除的常见标点符号

    exclude_symbols = r"[a-zA-Z0-9\s,.!?;:()\"\'/#%-|,’“”·$_—。，]"

    try:
        # 打开并读取文件内容
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()

        # 遍历内容并匹配不在排除范围内的字符
        special_chars = [(char, hex(ord(char)), pos)
                         for pos, char in enumerate(content)
                         if not re.match(exclude_symbols, char)]

        # 打印特殊字符及其 Unicode 编码
        for char, unicode_value, position in special_chars:
            print(f"特殊字符: '{char}' | Unicode: {unicode_value} | 位置: {position}")

    except FileNotFoundError:
        print(f"文件未找到: {file_path}")
    except Exception as e:
        print(f"出现错误: {e}")

# 示例文件路径
file_path = r'C:\Users\owner\Desktop\KG\OneNYC_2050_Strategic_Plan.txt'
find_uncommon_special_characters(file_path)
```

```python
import re

def combine_page_content(input_path, output_path):
    """
    将文本文件中的每个页面内容合并成一段话，排除包含页码和链接的行。

    :param input_path: 输入文件路径
    :param output_path: 输出文件路径
    """
    try:
        with open(input_path, 'r', encoding='utf-8') as file:
            lines = file.readlines()

        # 用于保存合并后的内容
        combined_text = []
        page_content = []

        # 正则表达式匹配页码分割行（例如: '9 | OneNYC 2050' 和 'NYC.GOV/OneNYC'）
        page_splitter_pattern = re.compile(r"^\d+\s*\|\s*OneNYC.*|^NYC\.GOV/OneNYC")

        for line in lines:
            # 检查是否为页码分割行或链接行
            if page_splitter_pattern.match(line.strip()):
                if page_content:  # 如果页面内容不为空，进行合并
                    # 合并页面内容并去掉换行符
                    combined_text.append(" ".join(page_content))
                    page_content = []  # 清空当前页面内容
            else:
                # 添加非页码分割行的内容到当前页面内容中
                page_content.append(line.strip())

        # 确保最后一页的内容也被合并
        if page_content:
            combined_text.append(" ".join(page_content))

        # 将合并后的内容写入新文件
        with open(output_path, 'w', encoding='utf-8') as file:
            file.write("\n".join(combined_text))

        print(f"合并后的文件已保存至：{output_path}")

    except FileNotFoundError:
        print(f"文件未找到: {input_path}")
    except Exception as e:
        print(f"出现错误: {e}")

# 示例使用
input_path = r'C:\Users\owner\Desktop\KG\OneNYC_2050_Strategic_Plan.txt'
output_path = r'C:\Users\owner\Desktop\KG\OneNYC_2050_Strategic_Plan_combined.txt'
combine_page_content(input_path, output_path)
```

```python
def merge_paragraphs(input_path, output_path):
    """
    将文本文件中的每个段落合并为一段话，去除空行。

    :param input_path: 输入文件路径
    :param output_path: 输出文件路径
    """
    try:
        with open(input_path, 'r', encoding='utf-8') as file:
            lines = file.readlines()

        # 用于保存合并后的段落内容
        merged_paragraphs = []
        paragraph = []

        for line in lines:
            # 如果是空行，则将当前段落合并，并清空
            if line.strip() == "":
                if paragraph:  # 如果段落不为空
                    merged_paragraphs.append(" ".join(paragraph))
                    paragraph = []  # 清空段落内容
            else:
                # 将当前行内容加入到段落中
                paragraph.append(line.strip())

        # 确保最后一个段落也被合并
        if paragraph:
            merged_paragraphs.append(" ".join(paragraph))

        # 将合并后的段落写入新文件
        with open(output_path, 'w', encoding='utf-8') as file:
            file.write("\n".join(merged_paragraphs))

        print(f"合并后的文件已保存至：{output_path}")

    except FileNotFoundError:
        print(f"文件未找到: {input_path}")
    except Exception as e:
        print(f"出现错误: {e}")

# 示例使用
input_path = r'C:\Users\owner\Desktop\KG\上海市城市总体规划 （ 2017-2035年）文本.txt'
output_path = r'C:\Users\owner\Desktop\KG\上海市城市总体规划（2017-2035年）文本_clean.txt'
merge_paragraphs(input_path, output_path)
```

```python
import re

def remove_page_numbers(input_path, output_path):
    """
    识别文本中的页码并将其替换为空行。

    :param input_path: 输入文件路径
    :param output_path: 输出文件路径
    """
    try:
        with open(input_path, 'r', encoding='utf-8') as file:
            lines = file.readlines()

        # 页码识别的正则表达式：匹配单独的数字或数字后跟点
        page_number_pattern = r"^\d+(\.\d+)?\s*$"

        # 处理每一行，如果匹配页码格式，则替换为空行
        processed_lines = []
        for line in lines:
            if re.match(page_number_pattern, line.strip()):
                processed_lines.append("\n")  # 将页码替换为空行
            else:
                processed_lines.append(line)  # 保留非页码行

        # 将处理后的内容写入新文件
        with open(output_path, 'w', encoding='utf-8') as file:
            file.writelines(processed_lines)

        print(f"处理后的文件已保存至：{output_path}")

    except FileNotFoundError:
        print(f"文件未找到: {input_path}")
    except Exception as e:
        print(f"出现错误: {e}")

# 示例使用
input_path = r'C:\Users\owner\Desktop\KG\上海市城市总体规划（2017-2035年）文本_clean.txt'
output_path = r'C:\Users\owner\Desktop\KG\上海市城市总体规划（2017-2035年）文本_clean_page.txt'
remove_page_numbers(input_path, output_path)
```

```python
import re

def add_empty_line_before_sections(input_path, output_path):
    """
    在文本中，匹配 "第X条"、"第X章"、"第X节" 等内容，并将这些内容独立为一行，保留其前后内容格式。
    对于 "第X章"、"第X节" 等格式的内容，后面添加换行,"第X章 \n xxx"。
    对于 "1、"、"一、"、"六、" 等格式，也添加空行。

    :param input_path: 输入文件路径
    :param output_path: 输出文件路径
    """
    try:
        with open(input_path, 'r', encoding='utf-8') as file:
            lines = file.readlines()

        # 匹配 "第X条" 形式的正则表达式
        section_pattern = r"(第([一二三四五六七八九十0-9]+)(条))"

        # 匹配 "第X章" 或 "第X节" 形式的正则表达式
        chapter_or_section_pattern = r"(第([一二三四五六七八九十0-9]+)(章|节))"

        # 匹配 "1、" 或 "一、" 等的数字和中文数字
        number_pattern = r"([一二三四五六七八九十0-9]+、)"

        # 处理每一行，按照要求将匹配的内容分行显示
        processed_lines = []
        for line in lines:
            # 匹配并替换 "第X条" 或 "第X章"/"第X节"，并添加换行
            line = re.sub(section_pattern, r"\n\1\n", line)
            line = re.sub(chapter_or_section_pattern, r"\n\1\n", line)
            line = re.sub(number_pattern, r"\n\1", line)
            processed_lines.append(line)

        # 将处理后的内容写入新文件
        with open(output_path, 'w', encoding='utf-8') as file:
            file.writelines(processed_lines)

        print(f"处理后的文件已保存至：{output_path}")

    except FileNotFoundError:
        print(f"文件未找到: {input_path}")
    except Exception as e:
        print(f"出现错误: {e}")

# 示例使用
input_path = r'C:\Users\owner\Desktop\KG\上海市城市总体规划（2017-2035年）文本_clean.txt'
output_path = r'C:\Users\owner\Desktop\KG\上海市城市总体规划（2017-2035年）文本_clean_final.txt'
add_empty_line_before_sections(input_path, output_path)
```
